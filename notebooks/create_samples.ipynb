```python
# scripts/create_samples.py (o simile)
import csv
import os

ORIGINAL_FILE = "../data/used_cars_cleaned.csv" # Adatta il percorso
SAMPLES_DIR = "../data/samples/"
os.makedirs(SAMPLES_DIR, exist_ok=True)

SAMPLE_SIZES = {
    "100": 100,
    "1k": 1000
}

try:
    with open(ORIGINAL_FILE, 'r', encoding='utf-8', errors='ignore') as infile: # errors='ignore' per robustezza iniziale
        reader = csv.reader(infile)
        header = next(reader)

        for name, size in SAMPLE_SIZES.items():
            output_filename = os.path.join(SAMPLES_DIR, f"used_cars_{name}.csv")
            with open(output_filename, 'w', newline='', encoding='utf-8') as outfile:
                writer = csv.writer(outfile)
                writer.writerow(header)
                print(f"Creating {output_filename}...")
                count = 0
                # Riapri il file originale per leggere i dati dopo l'header
                with open(ORIGINAL_FILE, 'r', encoding='utf-8', errors='ignore') as data_infile:
                    data_reader = csv.reader(data_infile)
                    next(data_reader) # Salta l'header
                    for i, row in enumerate(data_reader):
                        if i < size:
                            writer.writerow(row)
                            count +=1
                        else:
                            break
                print(f"Wrote {count} data lines to {output_filename}")
    print("Sample creation complete.")

except FileNotFoundError:
    print(f"Error: Original dataset file not found at {ORIGINAL_FILE}")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
import pandas as pd
import os

def main():
    # Definisci percorsi di input e output
    input_path = "../data/used_cars_cleaned.csv"
    output_dir = "../data/samples"
    output_path = os.path.join(output_dir, "used_cars_filtered.csv")
    
    # Assicurati che la directory di output esista
    os.makedirs(output_dir, exist_ok=True)
    
    # Definisci le colonne da mantenere per ciascun job
    JOB1_COLS = ['make_name', 'model_name', 'price', 'year']
    JOB2_COLS = ['city', 'year', 'price', 'daysonmarket', 'description', 'description_cleaned']
    
    # Combina le colonne e rimuovi i duplicati
    columns_to_keep = list(set(JOB1_COLS + JOB2_COLS))
    
    print(f"Lettura dataset da {input_path}...")
    # Leggi il dataset originale
    df = pd.read_csv(input_path)
    
    print(f"Dimensione dataset originale: {df.shape}")
    
    # Verifica se tutte le colonne richieste esistono nel dataset
    missing_cols = [col for col in columns_to_keep if col not in df.columns]
    if missing_cols:
        print(f"Attenzione: Le seguenti colonne richieste non sono presenti nel dataset: {missing_cols}")
        # Filtra per mantenere solo le colonne che esistono nel dataset
        columns_to_keep = [col for col in columns_to_keep if col in df.columns]
    
    # Filtra per mantenere solo le colonne richieste
    df_filtered = df[columns_to_keep]
    
    print(f"Dimensione dataset filtrato: {df_filtered.shape}")
    print(f"Colonne mantenute: {df_filtered.columns.tolist()}")
    
    # Salva il dataset filtrato
    print(f"Salvataggio dataset filtrato in {output_path}...")
    df_filtered.to_csv(output_path, index=False)
    
    print("Completato!")

if __name__ == "__main__":
    main()
```
