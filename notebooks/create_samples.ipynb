```python
# scripts/create_samples.py (o simile)
import csv
import os

ORIGINAL_FILE = "../data/used_cars_data.csv" # Adatta il percorso
SAMPLES_DIR = "../data/samples/"
os.makedirs(SAMPLES_DIR, exist_ok=True)

SAMPLE_SIZES = {
    "100": 100,
    "1k": 1000
}

try:
    with open(ORIGINAL_FILE, 'r', encoding='utf-8', errors='ignore') as infile: # errors='ignore' per robustezza iniziale
        reader = csv.reader(infile)
        header = next(reader)

        for name, size in SAMPLE_SIZES.items():
            output_filename = os.path.join(SAMPLES_DIR, f"used_cars_{name}.csv")
            with open(output_filename, 'w', newline='', encoding='utf-8') as outfile:
                writer = csv.writer(outfile)
                writer.writerow(header)
                print(f"Creating {output_filename}...")
                count = 0
                # Riapri il file originale per leggere i dati dopo l'header
                with open(ORIGINAL_FILE, 'r', encoding='utf-8', errors='ignore') as data_infile:
                    data_reader = csv.reader(data_infile)
                    next(data_reader) # Salta l'header
                    for i, row in enumerate(data_reader):
                        if i < size:
                            writer.writerow(row)
                            count +=1
                        else:
                            break
                print(f"Wrote {count} data lines to {output_filename}")
    print("Sample creation complete.")

except FileNotFoundError:
    print(f"Error: Original dataset file not found at {ORIGINAL_FILE}")
except Exception as e:
    print(f"An error occurred: {e}")
```
